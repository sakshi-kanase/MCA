```python
#larest number
def largest(a,b,c):
    if a>=b and a>=c:
        return a
    elif b>=a and b>=c:
        return b
    else:
        return c
a = int(input("enter number"))
b = int(input("enter number"))
c = int(input("enter number"))
print("laregest number",largest(a,b,c))
```

```python
#factorial number
def factorial(n):
    fact = 1
    for i in range(1,n+1):
        fact *=1
    return fact
n = int(input("enter number"))
if n<0:
    print("not fact number")
else:
    print(factorial(n))
```
